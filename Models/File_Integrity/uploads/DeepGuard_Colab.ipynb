## ✅ Step 3 — Unzip everything

```
import zipfile, os

print('Unzipping... please wait')
with zipfile.ZipFile('deepguard_complete.zip', 'r') as z:
    z.extractall('/content/deepguard')

# Show what was extracted
for root, dirs, files in os.walk('/content/deepguard'):
    level = root.replace('/content/deepguard', '').count(os.sep)
    indent = ' ' * 2 * level
    print(f'{indent}{os.path.basename(root)}/')
    if level < 3:
        subindent = ' ' * 2 * (level + 1)
        for f in files[:3]:
            print(f'{subindent}{f}')

print('\n✅ Unzip complete!')
```
